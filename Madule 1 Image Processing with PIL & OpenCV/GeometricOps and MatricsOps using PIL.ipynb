```python
import matplotlib.pyplot  as plt
from PIL import Image
import numpy as np
```

```python
def plot_image(image_1, image_2,title_1="Orignal",title_2="New Image"):
    plt.figure(figsize=(10,10))
    plt.subplot(1, 2, 1)
    plt.imshow(image_1,cmap="gray")
    plt.title(title_1)
    plt.subplot(1, 2, 2)
    plt.imshow(image_2,cmap="gray")
    plt.title(title_2)
    plt.show()
```

**Geometric Operations**

```python
image = Image.open("lenna.png")
plt.imshow(image)
plt.show()
```

# 1. Scaling

```python
width,hight = image.size
new_width = 2*width
new_Image = image.resize((new_width,hight))
plt.imshow(new_Image)
plt.show()
```

# 2. Rotation

```python
theta = 45
rotated_image = image.rotate(theta)
plt.imshow(rotated_image)
plt.show()
```

# **Mathematical Operations**
Before doing that we must convert image to numpy array

```python
image = np.array(image)
```

```python
# we can add the constant to the array pixels
image_1 = image + 50
plot_image(image,image_1)
```

```python
# We can also multiply every pixel's intensity value by a constant value.
image_2 = 2*image
plot_image(image,image_2)
```

```python
#We can add the two arrays of equal size, for that we create an array of random values
Noise = np.random.normal(0, 20, (hight, width, 3)).astype(np.uint8)
Noise.shape
new_image = image + Noise
plot_image(image,new_image)
```

```python
# also we can multiply them
new_image = image * Noise
plot_image(image,new_image)
```
