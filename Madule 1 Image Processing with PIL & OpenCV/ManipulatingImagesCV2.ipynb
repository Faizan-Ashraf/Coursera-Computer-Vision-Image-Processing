```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

### **Copying Image**

```python
baboon = cv2.imread("baboon.png")
plt.figure(figsize=(3,3))
plt.imshow(cv2.cvtColor(baboon, cv2.COLOR_BGR2RGB)) #converts BGR(default) to RGB which matplotlib expects
plt.show()
```

```python
A = baboon.copy()
id(baboon) == id(A)
A[:,:,:] = 0 #[height, width, colorChannel]
plt.figure(figsize=(10,10))

plt.subplot(1,2,1)  #(row,column,position)
plt.imshow(cv2.cvtColor(baboon, cv2.COLOR_BGR2RGB))

plt.subplot(1,2,2) #(row,column,position)
plt.imshow(cv2.cvtColor(A, cv2.COLOR_BGR2RGB))


plt.show()
```

### **Flipping Images**

```python
img = cv2.imread("cat.png")
plt.figure(figsize = (3,3))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

```python

w,h,c = img.shape
print('width, height, C', w, h, c)
```

```python
# flipcode = 0 flip horizontally around x-axis
# flipcode < 0 flip vertically around y-axis
# flipcode > 0  flip vertically and horizontally, flipping around both axes negative value


for i, flipcode in enumerate([1, 0, -1], 1):
    plt.subplot(1, 3, i)
    im_flip = cv2.flip(img, flipcode)
    plt.imshow(cv2.cvtColor(im_flip, cv2.COLOR_BGR2RGB))
    plt.title(f"flipcode = {flipcode}")
    plt.axis('off')  # Optional: hides axis ticks

plt.tight_layout()
plt.show()
```

```python
im_flip = cv2.rotate(img,0)
plt.imshow(cv2.cvtColor(im_flip,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
flip = {"ROTATE_90_CLOCKWISE":cv2.ROTATE_90_CLOCKWISE,
        "ROTATE_90_COUNTERCLOCKWISE":cv2.ROTATE_90_COUNTERCLOCKWISE,
        "ROTATE_180":cv2.ROTATE_180}
```

```python
for key, value in flip.items():
    plt.subplot(1,2,1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("orignal")
    plt.subplot(1,2,2)
    plt.imshow(cv2.cvtColor(cv2.rotate(img,value), cv2.COLOR_BGR2RGB))
    plt.title(key)
    plt.show()
```

### **Cropping Images**

```python
upper = 150
lower = 400
crop_top = img[upper: lower,:,:]
plt.figure(figsize=(10,10))
plt.imshow(cv2.cvtColor(crop_top, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
left = 150
right = 400
crop_horizontal = crop_top[: ,left:right,:]
plt.figure(figsize=(5,5))
plt.imshow(cv2.cvtColor(crop_horizontal, cv2.COLOR_BGR2RGB))
plt.show()
```

### **Changing Specific Pixels**

```python
start_point, end_point = (left, upper),(right, lower)
image_draw = np.copy(img)
cv2.rectangle(image_draw, pt1=start_point, pt2=end_point, color=(0, 255, 0), thickness=3)
plt.figure(figsize=(5,5))
plt.imshow(cv2.cvtColor(image_draw, cv2.COLOR_BGR2RGB))
plt.show()
```

### **Writing on Images**

```python
image_draw=cv2.putText(img=img,text='Stuff',org=(10,500),color=(255,255,255),fontFace=4,fontScale=5,thickness=2) #org (x, y)
plt.figure(figsize=(10,10))
plt.imshow(cv2.cvtColor(image_draw,cv2.COLOR_BGR2RGB))
plt.show()
```
